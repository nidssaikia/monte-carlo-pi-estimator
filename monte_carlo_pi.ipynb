```python
import numpy as np
import matplotlib.pyplot as plt

def estimate_pi(num_points=5000):
    # 1. Generate random (x, y) coordinates between -1 and 1
    x = np.random.uniform(-1, 1, num_points)
    y = np.random.uniform(-1, 1, num_points)

    # 2. Calculate distance from the origin (x^2 + y^2 <= 1)
    inside_circle = (x**2 + y**2) <= 1

    # 3. Calculate Pi estimate
    points_inside = np.sum(inside_circle)
    pi_estimate = 4 * (points_inside / num_points)

    print(f"Total Points: {num_points}")
    print(f"Points Inside Circle: {points_inside}")
    print(f"Estimated Pi: {pi_estimate}")
    print(f"Actual Pi: {np.pi}")
    print(f"Error: {abs(pi_estimate - np.pi):.5f}")

    # 4. Plot the results
    plt.figure(figsize=(7, 7))
    plt.scatter(x[inside_circle], y[inside_circle], color='blue', s=2, label='Inside Circle')
    plt.scatter(x[~inside_circle], y[~inside_circle], color='red', s=2, label='Outside Circle')

    plt.title(f"Monte Carlo Estimation of $\pi$\nEstimate: {pi_estimate} (N = {num_points})")
    plt.xlabel("X")
    plt.ylabel("Y")
    plt.legend(loc="upper right")
    plt.axis('equal')

    # Save image directly in Colab
    plt.savefig("pi_visualization.png")
    plt.show()

# Run the function
estimate_pi(num_points=5000)
```
